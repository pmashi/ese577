## 2) Residual Connection (10 pts)

Residual connections are a type of skip-connection that learn residual functions with reference to the layer inputs, instead of learning unreferenced functions. It is a critical idea that enabled the learning of very deep networks -- the intuition is that it is easier to optimize the residual mapping than to optimize the original, unreferenced mapping. (A detailed explanation about the training benefit is not required for this question; however, it is added at the end of the question if you're interested.)

Nowadays, residual connections are standard practice in all sorts of deep neural network architectures—feed-forward nets, convolutional nets, and transformers.

Concretely, the idea of residual connections is quite simple: it copies a block's input and adds it to the output of the block. Mathematically, that is:
$$\text{out} = x + \text{SomeLayers}(x) = x + F(x)\enspace.$$

### (Side note: How residual connections help the training)

1. **Easier Identity Learning:** Residual connections make it easier for the network to learn the identity function. For layers where the direct copying of the input is optimal, the layers can simply learn to adjust  towards zero, effectively making the layer a no-op. This is particularly beneficial when additional layers do not contribute to a better representation for a given task.

2. **Improved Gradient Flow:** During backpropagation, gradients have to travel back through the network to update each layer's weights. Residual connections create shortcuts for this gradient flow, allowing gradients to be directly propagated to earlier layers without passing through multiple levels of transformation. Mathematically, if you consider the derivative chain rule applied in backpropagation, the addition operation in the residual connection does not mitigate the magnitude of the gradient, as it involves a direct path for the gradient flow besides the deeper transformations.

3. **Mitigation of Vanishing Gradients:** With the direct paths provided by residual connections, even when the gradients through the deep layers of the network start to vanish, the gradients through the shortcut paths are preserved, maintaining a stronger gradient signal for the early layers. This effectively alleviates the vanishing gradient problem, as the network can still benefit from gradient information that has not been excessively diminished through deep layers of transformation.

A critical component of the computation within a transformer is the self-attention mechanism. Depicted below is one self-attention head.

<img src="https://jorge-a-mendez.github.io/files/ese577f25/hw7/one-head.png" width=30% class="center">

where `MatMul` denotes a matrix multiplication operation for weighted sum.

Suppose that we are given a sequence $\{x^{(i)}\}_{i=1}^n$ of $n$ tokens, where each token $x^{(i)}\in\mathbb{R}^d$ is a $d\times1$ vector and we create the matrix $X\in\mathbb{R}^{n\times d}$ by transposing each $x^{(i)}$ and stacking them vertically.

Our transformer head needs to learn three weight matrices: $W_q$, $W_k$, and $W_v$, all of which are of dimensions $d\times d_k$. These three weight matrices are used with our tokens in order to produce the query $Q$, key $K$, and value $V$ matrices, respectively.

Specifically, fo each token, we generate a $d_k$-dimensional query vector via $q_i=W_q^\top x^{(i)}$, and stack the transposed of all $n$ such query vectors vertically to get the $Q$ matrix:
$$Q = \begin{bmatrix}\begin{array}{c} q_1 \\ q_2 \\ \vdots \\ q_{n} \end{array}\end{bmatrix}\enspace.$$

A similar process goes for the key matrix $K$ and the value matrix $V$. 

We can write out the embedding/projection more compactly as:
$$Q = XW_q \qquad K = XW_k \qquad V=XW_v\enspace.$$

## 4) Transformer Sequence Averager (30 pts)

We will construct a transformer that computes the average over a sequence of values seen so far. The model receives a sequence of tokens. Each token is either a **content token**, which contains a number to include in the average, or a **trigger token**, which requests the average of all preceding content tokens. A trigger does not reset the average.

Each token is encoded with three features:
$$x^{(i)} = [\text{is\_trigger},\ \text{is\_content},\ n]^\top.$$
The first two features are binary indicators, and $n$ is $0$ for trigger tokens and the value for content tokens. For example:
- $x^{(i)} = [0, 1, 5]^\top$ is a content token with $n=5$
- $x^{(i)} = [1,0,0]^\top$ is a trigger token.

Given a sequence of input tokens $x^{(1)}, \ldots x^{(N)}$, the transformer should produce an output sequence of the same length $y^{(1)}, \ldots, y^{(n)}$. At positions where the input is a trigger token, the output should be the average over previous content tokens. At positions where the input is a content token, the output should be $0$. Assume that all sequences begin with a trigger token and that the average of an empty sequence is defined to be $0$.

The output sequence has the same length as the input. At a trigger position, output the average of **all previous content values**; at a content position, output $0$. For example:

$$
\underset{\underset{\text{trigger}}{\uparrow}}{\begin{bmatrix}
        1\\0\\0
    \end{bmatrix}} \hspace{0.5em} 
    \begin{bmatrix}
        0\\1\\5
    \end{bmatrix} \hspace{0.5em}
    \begin{bmatrix}
        0\\1\\3
    \end{bmatrix}\hspace{0.5em}
    \begin{bmatrix}
        1\\0\\0
    \end{bmatrix}\hspace{0.5em}
    \begin{bmatrix}
        0\\1\\2
    \end{bmatrix}\hspace{0.5em}
    \begin{bmatrix}
        1\\0\\0
    \end{bmatrix}
    \rightarrow \underset{\underset{\text{avg}([])}{\uparrow}}{0} \hspace{2em} 0 \hspace{2em} 0 \hspace{2em} \underset{\underset{\text{avg}([5,3])}{\uparrow}}{4} \hspace{2em} 0 \hspace{2em} \underset{\underset{\text{avg}([5,3,2])}{\uparrow}}{3.33}
$$

Our transformer learns projection matrices $W_Q\in\mathbb{R}^{d_k,3}, W_K\in\mathbb{R}^{d_k,3}, W_V\in\mathbb{R}^{d_v,3}$ that map raw tokens $x^{(i)}$ to queries $q^{(i)}=W_Q\cdot x^{(i)}$, keys $k^{(i)}=W_K\cdot x^{(i)}$, and values $v^{(i)}=W_V\cdot x^{(i)}$, respectively. For each query $q^{(i)}$, we will compute unnormalized self-attention scores $a^{(i)}=\mathrm{softmax}({q^{(i)}}^\top k^{(1)}, \ldots, {q^{(i)}}^\top k^{(N)})$ for all $N$ keys in the sequence. The $i$-th output will be computed as the sum of values weighted by the self-attention scores: $y^{(i)}=a^{(i)}_1 v^{(1)} + \cdots + a^{(i)}_N v^{(N)}$.

````python
def apply_sequence_averager_4(X, WQ, WK, WV, causal=True):
    """Run the simplified attention head defined in Question 4."""
    X = np.asarray(X, dtype=float)
    Q = X @ np.asarray(WQ, dtype=float).T
    K = X @ np.asarray(WK, dtype=float).T
    V = X @ np.asarray(WV, dtype=float).T
    scores = Q @ K.T
    if causal:
        scores[np.triu_indices(len(X), k=1)] = -np.inf
    weights = np.exp(scores - np.max(scores, axis=1, keepdims=True))
    attention = weights / weights.sum(axis=1, keepdims=True)
    return attention, attention @ V

A_example_4, Y_example_4 = apply_sequence_averager_4(
    X_example_4, WQ_414, WK_414, WV_422, causal=ans_411
)
np.testing.assert_allclose(
    Y_example_4, np.array([[0], [0], [0], [4], [0], [10/3]]),
    rtol=0, atol=1e-6,
)
print("Example outputs:", np.round(Y_example_4.ravel(), 6))
````

## 6) foodGPT (10 pts)

A transformer can be used as a chatbot by having it output one token at a time. The output token is then appended to the input, and the process is repeated. This is called an *autoregressive* configuration.

Here, we explore the use of the GPT model from recitation to make a very simple chatbot that generates recipe names. Our training data is the same set of 1200 recipe names used in the previous problem.

The model and code for training it is below. Read over this code and try to understand how it works, at a functional level.

````python
# hyperparameters
# max_iters = 1000   # use this if training from scratch
max_iters = 0      # use this if loading saved state
eval_interval = 100
learning_rate = 3e-4
device = 'cpu'
eval_iters = 200
batch_size = 16 # how many independent sequences will we process in parallel?
block_size = 32 # what is the maximum context length for predictions?

d = 64
n_head = 4
n_layer = 4
````

````python
!wget --no-check-certificate "https://drive.google.com/uc?export=download&id=1406zdBp0MiDSUe4PUMsoHAND5B7ir1j6" -O food.txt
!wget --no-check-certificate "https://drive.google.com/uc?export=download&id=1OcCFCgBwL_TZG1HaGk7yCUAa7Zho8C7l" -O foodGPT_state.pth

sfn = "foodGPT_state.pth"

torch.manual_seed(1337)

with open('food.txt', 'r', encoding='utf-8') as f:
    text = f.read()
    text = text.lower()

# here are all the unique characters that occur in this text
chars = sorted(list(set(text)))
vocab_size = len(chars)
# create a mapping from characters to integers and back
stoi = { ch:i for i,ch in enumerate(chars) }
itos = { i:ch for i,ch in enumerate(chars) }
encode = lambda s: [stoi[c] for c in s] # encoder: take a string, output a list of integers
decode = lambda l: ''.join([itos[i] for i in l]) # decoder: take a list of integers, output a string

# Train and test splits
data = torch.tensor(encode(text), dtype=torch.long)
n = int(0.9*len(data)) # first 90% will be train, rest val
# train_data = data[:n]
train_data = data
val_data = data[n:]

# data loading
def get_batch(split):
    # generate a small batch of data of inputs x and targets y
    data = train_data if split == 'train' else val_data
    ix = torch.randint(len(data) - block_size, (batch_size,))
    x = torch.stack([data[i:i+block_size] for i in ix])
    y = torch.stack([data[i+1:i+block_size+1] for i in ix])
    x, y = x.to(device), y.to(device)
    return x, y

@torch.no_grad()
def estimate_loss():
    out = {}
    model_6.eval()
    for split in ['train', 'val']:
        losses = torch.zeros(eval_iters)
        for k in range(eval_iters):
            X, Y = get_batch(split)
            logits, loss = model_6(X, Y)
            losses[k] = loss.item()
        out[split] = losses.mean()
    model_6.train()
    return out
````

````python
# GPT model

class Head(nn.Module):
    """ one head of self-attention """

    def __init__(self, d_k):
        super().__init__()
        self.key = nn.Linear(d, d_k, bias=False)
        self.query = nn.Linear(d, d_k, bias=False)
        self.value = nn.Linear(d, d_k, bias=False)
        self.register_buffer('tril', torch.tril(torch.ones(block_size, block_size)))

    def forward(self, x):
        # input of size (batch, time-step, channels)
        # output of size (batch, time-step, head size)
        B,T,C = x.shape
        k = self.key(x)   # (B,T,hs)
        q = self.query(x) # (B,T,hs)
        # compute attention scores ("affinities")
        wei = q @ k.transpose(-2,-1) * k.shape[-1]**-0.5 # (B, T, hs) @ (B, hs, T) -> (B, T, T)
        wei = wei.masked_fill(self.tril[:T, :T] == 0, float('-inf')) # (B, T, T)
        wei = F.softmax(wei, dim=-1) # (B, T, T)
        # perform the weighted aggregation of the values
        v = self.value(x) # (B,T,hs)
        out = wei @ v # (B, T, T) @ (B, T, hs) -> (B, T, hs)
        return out

class MultiHeadAttention(nn.Module):
    """ multiple heads of self-attention in parallel """

    def __init__(self, num_heads, d_k):
        super().__init__()
        self.heads = nn.ModuleList([Head(d_k) for _ in range(num_heads)])

    def forward(self, x):
        out = torch.cat([h(x) for h in self.heads], dim=-1)
        return out

class FeedFoward(nn.Module):
    """ a simple linear layer followed by a non-linearity """

    def __init__(self, d):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(d, 4 * d),
            nn.ReLU(),
            nn.Linear(4 * d, d),
        )

    def forward(self, x):
        return self.net(x)

class Block(nn.Module):
    """ Transformer block: communication followed by computation """

    def __init__(self, d, n_head):
        # d: embedding dimension, n_head: the number of heads we'd like
        super().__init__()
        d_k = d // n_head
        self.sa = MultiHeadAttention(n_head, d_k)
        self.ffwd = FeedFoward(d)
        self.ln1 = nn.LayerNorm(d)
        self.ln2 = nn.LayerNorm(d)

    def forward(self, x):
        x = x + self.sa(self.ln1(x))
        x = x + self.ffwd(self.ln2(x))
        return x

class GPTLanguageModel(nn.Module):

    def __init__(self):
        super().__init__()
        # each token directly reads off the logits for the next token from a lookup table
        self.token_embedding_table = nn.Embedding(vocab_size, d)
        self.position_embedding_table = nn.Embedding(block_size, d)
        self.blocks = nn.Sequential(*[Block(d, n_head=n_head) for _ in range(n_layer)])
        self.ln_f = nn.LayerNorm(d) # final layer norm
        self.lm_head = nn.Linear(d, vocab_size)

        # better init, not covered in the original GPT video, but important, will cover in followup video
        self.apply(self._init_weights)

    def _init_weights(self, module):
        if isinstance(module, nn.Linear):
            torch.nn.init.normal_(module.weight, mean=0.0, std=0.02)
            if module.bias is not None:
                torch.nn.init.zeros_(module.bias)
        elif isinstance(module, nn.Embedding):
            torch.nn.init.normal_(module.weight, mean=0.0, std=0.02)

    def forward(self, idx, targets=None):
        B, T = idx.shape

        # idx and targets are both (B,T) tensor of integers
        tok_emb = self.token_embedding_table(idx) # (B,T,C)
        pos_emb = self.position_embedding_table(torch.arange(T, device=device)) # (T,C)
        x = tok_emb + pos_emb # (B,T,C)
        x = self.blocks(x) # (B,T,C)
        x = self.ln_f(x) # (B,T,C)
        logits = self.lm_head(x) # (B,T,vocab_size)

        if targets is None:
            loss = None
        else:
            B, T, C = logits.shape
            logits = logits.view(B*T, C)
            targets = targets.view(B*T)
            loss = F.cross_entropy(logits, targets)

        return logits, loss

    def generate(self, idx, max_new_tokens, end_token=None):
        '''
        idx is (B, T) array of indices in the current context
        If end_token is provided, then terminate when that is generated
        '''
        for _ in range(max_new_tokens):
            # crop idx to the last block_size tokens
            idx_cond = idx[:, -block_size:]
            # get the predictions
            logits, loss = self(idx_cond)
            # focus only on the last time step
            logits = logits[:, -1, :] # becomes (B, C)
            # apply softmax to get probabilities
            probs = F.softmax(logits, dim=-1) # (B, C)
            # sample from the distribution
            idx_next = torch.multinomial(probs, num_samples=1) # (B, 1)
            if end_token is not None and idx_next==end_token:
                break
            # append sampled index to the running sequence
            idx = torch.cat((idx, idx_next), dim=1) # (B, T+1)
        return idx
````
