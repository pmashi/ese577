# Logistic regression

## 1) Transferring a learned logistic classifier

Suppose we use logistic regression on a dataset derived from people living in New York to learn a logistic classifier $\sigma(\theta^\top x+\theta_0)$ giving the probability that an adult with features $x$ will develop a heart disease in the next decade. Let $\theta^*$ and $\theta_0^*$ be the set of parameters that we learn.

We would like to use the same logistic regression classifier (with the $\theta^*$ and $\theta_0^*$ learned for New York residents) to make predictions for people living in Norway. However, we notice that heart disease is much less common in Norway and think that the model may need to be adjusted to account for this. For now, we decide to continue using logistic regression and decide to consider only changes to the offset $\theta_0$ (that is, we continue using the same $\theta=\theta^*$ from New York).

### a)
Consider a specific patient with feature vector $x$. How should we adjust the offset parameter, relative to $\theta_0^*$ learned in New York, so as to lower the predicted probability of this patient developing heart disease?

### b)

We realize that choosing the right value of $\theta_0$ is tricky since we don't have any labeled data from Norway. We decide to only use the model to find the $10\%$ individuals with the highest probability of developing heart disease so that we can closely follow them and make sure they are tested appropriately. Is the following statement true?: "We can use any value of $\theta_0$ and the original values of $\theta^*$."
