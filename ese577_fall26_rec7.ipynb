## 2)

We have a regression problem with several features. We have encoded the features so that some are continuous and in the range $[0, 1]$, while others are discrete and can only take on values $0$ or $1$.

Ava worries that having some features in a continuous range $[0, 1]$ and others in a discrete range $\{0, 1\}$ would mean that regularization will not work well to build regression models.

Ben argues that with Ava's feature encodings ensuring that features are between 0 and 1, regularization will no longer be needed.

Carla is most interested in analysis of the importance of different features and believes that these encodings will make that difficult.


We decide to proceed with our encodings and to include a ridge $\lambda\|\theta\|^2$ regularization term in our objective. Why, and how do we respond to Ava, Ben, and Carla?
